Python tem 4 tipos primitivos principais

int=números inteiros (7,-4,0)

float=números reais (4.5,0.078,-15.223) ponto flutuante

bool=valores lógicos ou booleanos (True,False) verdadeiro ou falso

str= valores de caracteres ('ola')
